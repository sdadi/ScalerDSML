```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from scipy.stats import norm
from scipy.stats import power
from scipy.stats import ttest_1samp
import os
```

### One Sample T testt

```python
# H0 : pop_mean = 100
# H1 : pop_mean > 100
pop_mean = 100
iq_scores = [110, 105, 98, 102, 99, 104, 115, 95]
t_stat, p_value = ttest_1samp(iq_scores, pop_mean)
print(f"T-statistic: {t_stat}, P-value: {p_value}")

alpha=0.01

if p_value < alpha:
    print("Reject the H0 - NULL hypothesis: The average IQ score is significantly different from 100.")
else:
    print("Fail to reject the H0 - NULL hypothesis: The average IQ score is not significantly different from 100.")
```

Two Sample T test

```python
script_dir = os.path.dirname(os.path.abspath('__file__'))
print(script_dir)
df_iq = pd.read_csv('D:\\learning\\ScalerDSML\\assets\\iq_two_schools.csv')
df_iq.head()
```

```python
df_iq.value_counts()
```

```python
df_iq.groupby('School')['iq'].mean()
```

### H0 u1 = u2
### HA u1 != u2

```python
from scipy.stats import ttest_1samp ## This is used for 1 sample
from scipy.stats import ttest_ind ## This is used for 2 sample
```

```python
iq_1 = df_iq[df_iq['School'] == 'school_1']['iq']
iq_2 = df_iq[df_iq['School'] == 'school_2']['iq']

t_stat, p_value = ttest_ind(iq_1, iq_2, alternative="two-sided")
print(f"T-statistic: {t_stat}, P-value: {p_value}")

alpha = 0.05 # 95% confidence
if p_value < alpha:
    print("Reject the H0 - NULL hypothesis")
else:
    print("Fail to reject the H0 - NULL hypothesis")
```

### Paired T Test

```python
from scipy.stats import ttest_rel
```

```python
df_ps = pd.read_csv('D:\\learning\\ScalerDSML\\assets\\problem_solving.csv')
df_ps.shape
```

```python
df_ps.head()
```

### H0: No Impact (marks remain same), Problem Solving has no effect
### HA: Some effect (marks are not same), Problem solving has an effect

### Two tail test (H0 != HA)

```python
statistic, p_value = ttest_rel(df_ps['test_1'], df_ps['test_2'])
print(f"T-statistic: {statistic}, P-value: {p_value}")

alpha = 0.05
if p_value < alpha:
    print(f"{p_value} < {alpha}")
    print("Reject the H0 - NULL hypothesis")
else:
    print(f"{p_value} >= {alpha}")
    print("Fail to reject the H0 - NULL hypothesis")
```

# H0: No effect
# HA: Problem solving improved the scores (u1 < u2)

```python
statistic, p_value = ttest_rel(df_ps['test_1'], df_ps['test_2'], alternative='less')
print(f"T-statistic: {statistic}, P-value: {p_value}")

alpha = 0.05
if p_value < alpha:
    print(f"{p_value} < {alpha}")
    print("Reject the H0 - NULL hypothesis")
else:
    print(f"{p_value} >= {alpha}")
    print("Fail to reject the H0 - NULL hypothesis")
```
