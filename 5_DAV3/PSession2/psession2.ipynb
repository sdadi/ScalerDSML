```python
import pandas as pd
url ="https://d2beiqkhq929f0.cloudfront.net/public_assets/assets/000/133/892/original/Ride_Sharing_Service.csv?1750314893"
df=pd.read_csv(url)
df.head()
```

```python
df['tip_percent'] = (df['tip_amount']/df['fare_amount'])*100
df['is_generous'] = (df['tip_percent'] > 5).astype(int)
print('count of generous rides:', df[df['is_generous'] == 1].shape[0])

generous_ride_types = df.groupby('ride_type')['is_generous'].mean().sort_values(ascending=False)
print(generous_ride_types)
df[df['is_generous'] == 1].value_counts('ride_type')
```
