### Q1:

When smokers smoke, nicotine is transformed into cotinine, which can be tested.
The average cotinine level in a group of 50 smokers was 243.5 ng ml.
Assuming that the standard deviation is known to be 229.5 ng ml.
Test the assertion that the mean cotinine level of all smokers is equal to 300.0 ng ml, at 95% confidence.

A) P-value: 0.0408, the mean cotinine level of all smokers is not equal to 300.0 ng/ml

B)P-value: 0.0408 , the mean cotinine level of all smokers is equal to 300.0 ng/ml.

C)P-value: 0.0817 , the mean cotinine level of all smokers is not equal to 300.0 ng/ml

D)P-value: 0.0817 , the mean cotinine level of all smokers is equal to 300.0 ng/ml.

```python
# H0: Average cotinine level equals 300.0 ng/ml
# HA: Average cotinine level not equals 300.0 ng/ml

sample_size = 50
sample_mean = 243.5
std_deviation = 229.5
population_mean = 300.0

confidence_level = 0.95
alpha = 1 - confidence_level

std_error = std_deviation / np.sqrt(sample_size)
print("Standard Error:", std_error)
z=(sample_mean - population_mean) / std_error
print("Z-score:", z)


p_value = norm.cdf(z)
print("One side P-value:", p_value)
final_p_value = 2*p_value
print("Final P-value:", final_p_value)

if final_p_value < alpha:
    print("Reject the null hypothesis: The average cotinine level is not equal to 300.0 ng/ml.")
else:
    print("Fail to reject the null hypothesis: The average cotinine level is equal to 300.0 ng/ml.")
```

## Q2:

There is a website, "www.goodreads.com" where people can leave their reviews about a book and rate them on a scale of 1 to 5 stars.
The following table gives the number of pages of a random sample of books with different ratings, as listed on the site:
one_star = [382, 391, 335, 368, 400, 372] two_star = [560, 343, 512, 329, 391, 367] three_star = [384, 458, 409, 309, 374, 459] four_star =
[325, 390, 304, 240, 306, 169] five_star = [360, 298, 272, 368, 320, 326]
At a 1% significance level, does the data provide sufficient evidence to conclude that there is a difference in the mean number of pages
among books in these five rating groups?
Choose the correct option below :

A) p_value = 0.03121, Number of pages in the books belonging to different rating groups is different

B) p_value = 0.03121, Number of pages in the books belonging to different rating groups is same

C) p_value = 0.00936, Number of pages in the books belonging to different rating groups is different

D) p_value = 0.00936, Number of pages in the books belonging to different rating groups is same

```python
# H0: No diff
# HA: There is a significant difference in average ratings among the groups.

one_star = [382, 391, 335, 368, 400, 372] 
two_star = [560, 343, 512, 329, 391, 367] 
three_star = [384, 458, 409, 309, 374, 459] 
four_star = [325, 390, 304, 240, 306, 169] 
five_star = [360, 298, 272, 368, 320, 326]

alpha = 0.01

f_statistic, p_value = f_oneway(one_star, two_star, three_star, four_star, five_star)
print("F-statistic:", f_statistic)
print("P-value:", p_value)

if p_value < alpha:
    print("Reject the null hypothesis: There is a significant difference in average ratings among the groups.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in average ratings among the groups.")

print("Answer is Option:C")
```

##### Q3:
In a Walmart store, there are two types of self-checkout machines: Type A and Type B. Type A machines have a higher accuracy rate of correctly scanning items compared to Type B machines. However, Type B machines are more popular among customers due to their faster processing speed.
#### Question:
Suppose a customer is observed using a self-checkout machine, and it is known that the probability of the customer using a Type A
machine is 0.3, while the probability of the customer using a Type B machine is 0.7. Additionally, it is known that the probability of a
Type A machine correctly scanning an item is 0.95, and the probability of a Type B machine correctly scanning an item is 0.90. If a
customer's item is observed to be scanned correctly, what is the probability that the customer used a Type A machine?



```python
type_a = 0.3
type_b = 0.7

a_scan = 0.95
b_scan = 0.9

total_scan = type_a * a_scan + type_b * b_scan
print("Total Scan:", total_scan)

scanned_a = (type_a * a_scan) / total_scan
print("Scanned A:", scanned_a)
```
