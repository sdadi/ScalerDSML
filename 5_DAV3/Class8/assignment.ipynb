### Q1. Dosage and Gender</br>
A researcher is interested in determining the effects of different dosages of a dietary supplement on the performance of both males and females on a physical endurance test.</br>
The three different dosages of the medicine are low, medium, and high, and the genders are male and female.</br>

Conduct an appropriate hypothesis test to determine the interaction effects of the test at a 1% significance level.

```python
data = """Dietary,Supplement_Dosage,Test_values
Female,Low,35.6
Female,Medium,49.4
Female,High,55.2
Male,Low,92.2
Male,Medium,45.4
Male,High,70.8
Female,Low,56.9
Female,Medium,87.0
Female,High,23.4
Male,Low,45.9
Male,Medium,34.2
Male,High,98.9"""
df = pd.read_csv(io.StringIO(data))

df.head()
```

```python
model = ols('Test_values ~ C(Supplement_Dosage)*C(Dietary)', data=df).fit()
anova_table = sm.stats.anova_lm(model, type=2)
print(f'There is no significant difference in average performance on the physical endurance test based on gender, supplement dosage, or their interaction.')
anova_table
```

### Q2. Fertiliser and Watering Frequency <br/>
A researcher wants to investigate the effects of two different fertilizers (‘A’ & ‘B’) and three watering frequencies (‘Low’, ‘Medium’, ‘High’) on the growth of tomato plants.<br/>

Is there a significant interaction between the fertilizer type and watering frequency on plant growth?<br/>

Conduct an appropriate hypothesis test to determine only the interaction effects of this research at a 5% significance level.

```python
data = """Fertilizer,Watering_Frequency,Plant_Height
A,Low,15.2
A,Medium,20.7
A,High,24.3
B,Low,18.4
B,Medium,23.1
B,High,26.5
A,Low,32.7
A,Medium,21.7
A,High,27.3
B,Low,38.4
B,Medium,33.1
B,High,36.5
A,Low,12.2
A,Medium,22.7
A,High,35.3
B,Low,28.4
B,Medium,22.1
B,High,24.5"""
df = pd.read_csv(io.StringIO(data))
df.head()
```

### Q3. Education vs Test
The Committee head of a national entrance exam wants to analyze if there are any differences in learning outcomes between students with different educational backgrounds (high school or college) and teaching methods (traditional or interactive) on test scores.

Conduct an appropriate hypothesis test to determine the main effects & interaction effects of the test at a 5% significance level.

```python
data = """Education,Teaching_Method,Test_Score
High School,Traditional,72
High School,Interactive,85
College,Traditional,70
College,Interactive,92
High School,Traditional,74
High School,Interactive,97
College,Traditional,76
College,Interactive,94
High School,Traditional,86
High School,Interactive,90
College,Traditional,69
College,Interactive,78
High School,Traditional,82
High School,Interactive,93
College,Traditional,98
College,Interactive,92"""
df = pd.read_csv(io.StringIO(data))
df.head()
```

```python
model = ols('Test_Score ~ C(Education)*C(Teaching_Method)',data=df).fit()
anova_table = sm.stats.anova_lm(model,type=2)
print(f'The teaching method significantly impacts student test scores, whereas educational background does not.')
print(f'There is no significant interaction between educational background and teaching method on student performance.')
anova_table
```

### Q4. Sales Strategy
A data analyst is comparing the sales amounts (in dollars) for two different marketing strategies (A and B). The sales data for 20 days under each strategy is collected.


sales_strategy_A = [156, 153, 157, 154, 156, 159, 152, 156, 157, 154, 153, 157, 157,152, 155, 154, 151, 157, 155, 151]<br/>
sales_strategy_B = [135, 147, 126, 136, 158, 139, 163, 141, 156, 142, 130, 129, 161, 158, 117, 151, 121, 135, 123, 153]<br/>

Perform an appropriate test to assess if there is a significant difference in the sales distributions between Strategy A and Strategy B. Use a significance level of 0.05.

```python
H0 = 'Strategy A and B are no significant difference'
HA = 'Strategy A and B are significant different'

sales_strategy_A = [156, 153, 157, 154, 156, 159, 152, 156, 157, 154, 153, 157, 157,152, 155, 154, 151, 157, 155, 151]
sales_strategy_B = [135, 147, 126, 136, 158, 139, 163, 141, 156, 142, 130, 129, 161, 158, 117, 151, 121, 135, 123, 153]
alpha = 0.05

stat, p_value = kstest(sales_strategy_A,sales_strategy_B)
print(f'stat is: {stat}, p_value:{p_value:.8f}')

if p_value < alpha:
    print(f'Reject HO means: {HA}')
else:
    print(f'Fail to Reject HO means: {H0}')
```

### Q7. Banking app
A bank is launching two different approaches (A and B) to encourage customers to adopt its new mobile banking app. The bank randomly assigns a group of customers to each approach and monitors their adoption rates over a month.<br/>

Data:<br/>

Group A (Approach A): [38, 40, 42, 37, 39, 41, 36, 35, 43, 38] <br/> 
Group B (Approach B): [48, 45, 46, 43, 50, 44, 49, 47, 42, 46] <br/>

Note: Assume both Group A & Group B follow the normal distribution.<br/>
Objective:<br/>

Assess whether the new incentive program in Approach B leads to a statistically significant improvement in the average adoption rates compared to Approach A.

Choose a suitable statistical test.

```python
H0 = 'Groups A and B are no significant improvement'
HA = 'Groups B are significant improvement over Group A'

groupA = [38, 40, 42, 37, 39, 41, 36, 35, 43, 38]  
groupB = [48, 45, 46, 43, 50, 44, 49, 47, 42, 46] 

alpha = 0.05

stat, p_value = ttest_ind(groupA,groupB,alternative='less')
print(f'stat is: {stat}, p_value:{p_value:.10f}')

if p_value < alpha:
    print(f'Reject HO means: {HA}')
else:
    print(f'Fail to Reject HO means: {H0}')
```
