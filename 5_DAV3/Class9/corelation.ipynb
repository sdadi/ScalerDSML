```python
df_hw = pd.read_csv('../../assets/weight-height.csv')
df_hw.head()
```

```python
sns.scatterplot(x=df_hw['Height'], y=df_hw['Weight'])
```

```python
p_cor = df_hw[['Height','Weight']].corr()
print(f'Pearson corelation is:')
p_cor
```

<ul>
<li>Ranked Data is also called as Ordinal data, whereas un-ranked data is called Nominal data</li>
<li>This unique type of correlation, driven by ranks, is called the Spearman Correlation.</li>
<ul><li>It's perfect for situations where relationships are more complex than what linear patterns can capture.</li></ul>
<li>The <b>Spearman Correlation</b> helps us uncover connections that might be hidden when using linear measures.</li>
</ul>

```python
# H0: No correlation
# Ha: There is correlation
pearsonr(df_hw["Height"], df_hw["Weight"])
```

```python
# H0: No correlation
# Ha: There is correlation
spearmanr(df_hw["Height"], df_hw["Weight"])
```

```python
df = pd.read_csv('../../assets/concrete_data.csv')
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(font_scale=1.0)
plt.figure(figsize=(10,8))
correlation_values = df.corr(method = 'pearson')
sns.heatmap(correlation_values, vmax = .6, linewidths=0.01, square=True, annot=True, cmap="RdBu", linecolor="black")
plt.title('Correlation between features')
```
