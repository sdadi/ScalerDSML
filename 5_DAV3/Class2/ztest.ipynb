```python
import pandas as pd
import numpy as np
from scipy.stats import norm
from scipy.stats import power
```

#### Firm A

```python
mu = 1800 # average sales
sigma = 100 # standard deviation
sample_size = 50 # sample size
alpha = 0.01 # significance level

z = (1850 - mu) / (sigma / np.sqrt(sample_size)) # Z-score calculation for sample mean of 1850
p_value = 1 - norm.cdf(z) # P-value for one(right)-tailed test
print (f'Z-score: {z}, P-value: {p_value}')

if(p_value < alpha):
    print("Reject H0: Reject the null hypothesis")
else:
    print("Accept H0: Fail to reject the null hypothesis")
```

#### What is the minimum weekly average sales below which I can say the Marketing Advertisement is not working
Second method to validate Firm A using critical value method

```python
mu = 1800 # average sales
sigma = 100 # standard deviation
sample_size = 50 # sample size
alpha = 0.01 # significance level

z = (1850 - mu) / (sigma / np.sqrt(sample_size)) # Z-score calculation for sample mean of 1850
p_value = 1 - norm.cdf(z) # P-value for one(right)-tailed test
print (f'Z-score: {z}, P-value: {p_value}')

if(p_value < alpha):
    print("Reject H0: Reject the null hypothesis")
else:
    print("Accept H0: Fail to reject the null hypothesis")

# Find the critical value of sample mean at alpha = 0.01  
# p_value = alpha
# 1-norm.cdf(z) == 0.01

z_critical = norm.ppf(1 - alpha)
print(f'Z-critical value for alpha {alpha} is: {z_critical}')

# z = (x-mu)/sigma
x = (z_critical*(sigma/np.sqrt(sample_size))) + mu
print(f'Critical value of sample mean at alpha {alpha} is: {x}')
```

### A fitness App claims that its users walk an avg of 8,000 steps per day.
### A random sample of 30 users showed an average of 7,600 steps per day with std deviation of 1,200 steps
### Conduct a left-tailed Z-test at a 5% significance level to determine if the Apps claim is supported. what is the p-value?


```python
mu = 8000 # average sales
sigma = 100 # standard deviation
sample_size = 30 # sample size
sample_mu = 7600
sample_std_dev= 1200
alpha = 0.05 # significance level

z = (sample_mu - mu) / (sample_std_dev / np.sqrt(sample_size)) # Z-score calculation for sample mean of 7600
p_value = norm.cdf(z) # P-value for one(right)-tailed test
print (f'Z-score: {z}, P-value: {p_value}')

if(p_value < alpha):
    print("Reject H0: Reject the null hypothesis")
else:
    print("Accept H0: Fail to reject the null hypothesis")

# Find the critical value of sample mean at alpha = 0.01  
# p_value = alpha
# 1-norm.cdf(z) == 0.01

z_critical = norm.ppf(1 - alpha)
print(f'Z-critical value for alpha {alpha} is: {z_critical}')

# z = (x-mu)/sigma
x = (z_critical*(sigma/np.sqrt(sample_size))) + mu
print(f'Critical value of sample mean at alpha {alpha} is: {x}')

```

Using Confidence Interval

```python
# H0: u = 1800
# HA: u > 1800

pop_mean = 1800
sample_mean = 1850
pop_std_dev = 100
sample_size = 50
alpha = 0.01

z = norm.ppf(1 - alpha)
print(f'Z-critical value for alpha {alpha} is: {z}')

marg_err = z * (pop_std_dev/np.sqrt(sample_size))
print(f'Margin of error is: {marg_err}')

confidence_interval = (sample_mean - marg_err, sample_mean + marg_err)
print(f'Confidence Interval is: {confidence_interval}')

if(confidence_interval[0] <= pop_mean <= confidence_interval[1]):
    print("Accept H0: Fail to reject the null hypothesis")
else:
    print("Reject H0: Reject the null hypothesis")

```
