### Q1. Energy Consumption
The United States is divided into four regions: Northeast, Midwest, South and West regions.<br/>
Independent random samples of households in these regions yielded the data on last year’s energy consumptions.

a = [13, 8, 11, 12, 11] <br/>
b = [15, 10, 16, 11, 13, 10]<br/>
c = [5, 11, 9, 5]<br/>
d = [8, 10, 6, 5, 7]<br/>

At a 5% significance level, does the data provide sufficient evidence to conclude that there is a difference in last year’s mean energy consumption
by households among the four regions?

```python
H0 = 'mean energy is similar'
HA = 'mean energy is different'

ne = [13, 8, 11, 12, 11]
mw = [15, 10, 16, 11, 13, 10]
s = [5, 11, 9, 5]
w = [8, 10, 6, 5, 7]

f_stat, p_value = f_oneway(ne, mw, s, w)
print(f'F_Oneway :{f_stat}')
print(f'p_value : {p_value}')

alpha = 0.05
if p_value < alpha:
    print(f'Reject H0 means: {HA}')
else:
    print(f'Fail to reject H0 means: {H0}')
```

### Q2. Book Review<br/>
There is a website, "www.goodreads.com" where people can leave their reviews about a book and rate them on a scale of 1 to 5 stars.<br/>
The following table gives the number of pages of a random sample of books with different ratings, as listed on the site:<br/>

one_star = [382, 391, 335, 368, 400, 372]<br/>
two_star = [560, 343, 512, 329, 391, 367]<br/>
three_star = [384, 458, 409, 309, 374, 459]<br/>
four_star = [325, 390, 304, 240, 306, 169]<br/>
five_star = [360, 298, 272, 368, 320, 326]<br/>

At a 1% significance level, does the data provide sufficient evidence to conclude that there is a difference in the mean number of pages among books in these five rating groups?

```python
H0 = 'Mean number of pages are similar'
HA = 'Mean number of pages are different'

one_star = [382, 391, 335, 368, 400, 372]
two_star = [560, 343, 512, 329, 391, 367]
three_star = [384, 458, 409, 309, 374, 459]
four_star = [325, 390, 304, 240, 306, 169]
five_star = [360, 298, 272, 368, 320, 326]

alpha = 0.01

f_stat, p_value = f_oneway(one_star, two_star, three_star, four_star, five_star)
print(f'F_One way stat: {f_stat}')
print(f'p_value is: {p_value}')

if p_value < alpha:
    print(f'Reject H0 means: {HA}')
else:
    print(f'Fail to reject H0 means: {H0}')
```

### Q3. SmartPhone Battery Life <br/>
Consumer Reports publishes reviews and comparisons of products based on results from its laboratory.<br/>
Data from their website gave the following table for battery lives in hours, for samples of smartphones made by four different mobile companies.<br/>

Brand_A = [19.60, 18.82, 19.00, 18.45, 19.79, 19.03, 17.89, 19.42]<br/>
Brand_B = [21.10, 20.00, 20.43, 19.67, 18.99, 19.98, 20.14, 19.78]<br/>
Brand_C = [10.31, 10.02, 9.41, 9.89, 10.05, 10.52, 11.02, 10.42]<br/>
Brand_D = [17.02, 16.71, 17.78, 18.65, 15.98, 17.63, 17.00, 16.78, 16.92, 17.14]<br/>

At a 2% significance level, does the data provide sufficient evidence to conclude that there is a significant difference in the mean battery life, among the four brands?<br/>

And, if significant, then perform pairwise T-tests to identify which pairs of smartphone brands offer a different mean battery life.

```python
H0 = 'Mean battery life is same among brands'
HA = 'Mean battery life is different among brands'
alpha = 0.02
Brand_A = [19.60, 18.82, 19.00, 18.45, 19.79, 19.03, 17.89, 19.42]
Brand_B = [21.10, 20.00, 20.43, 19.67, 18.99, 19.98, 20.14, 19.78]
Brand_C = [10.31, 10.02, 9.41, 9.89, 10.05, 10.52, 11.02, 10.42]
Brand_D = [17.02, 16.71, 17.78, 18.65, 15.98, 17.63, 17.00, 16.78, 16.92, 17.14]

f_stat, p_value = f_oneway(Brand_A, Brand_B, Brand_C, Brand_D)
print(f'F_One way stat: {f_stat}')
print(f'p_value is: {p_value}')

if p_value < alpha:
    print(f'Reject H0 means: {HA}')
else:
    print(f'Fail to reject H0 means: {H0}')

t_stat, p_value = ttest_ind(Brand_A, Brand_B)
print(' -----Brand A and B -----')
print(f'F_One way stat: {f_stat}')
print(f'p_value is: {p_value}')

if p_value < alpha:
    print(f'Reject H0 means: {HA}')
else:
    print(f'Fail to reject H0 means: {H0}')

t_stat, p_value = ttest_ind(Brand_C, Brand_D)
print(' -----Brand C and D -----')
print(f'F_One way stat: {f_stat}')
print(f'p_value is: {p_value}')

if p_value < alpha:
    print(f'Reject H0 means: {HA}')
else:
    print(f'Fail to reject H0 means: {H0}')

t_stat, p_value = ttest_ind(Brand_A, Brand_D)
print(' -----Brand A and D -----')
print(f'F_One way stat: {f_stat}')
print(f'p_value is: {p_value}')

if p_value < alpha:
    print(f'Reject H0 means: {HA}')
else:
    print(f'Fail to reject H0 means: {H0}')
```

### Q5. Same execution times?<br/>
A software development team is comparing the performance of three different algorithms for sorting large datasets.<br/>

The team collects the execution time (in milliseconds) for each algorithm on 20 different datasets:<br/>

algorithm_A = np.array([23, 25, 22, 27, 28, 24, 26, 29, 21, 30, 25, 43, 26, 28, 24, 22, 27, 46, 25, 29]) <br/>

algorithm_B = np.array([31, 28, 29, 32, 30, 33, 27, 28, 32, 30, 31, 29, 30, 48, 33, 31, 29, 30, 32, 31]) <br/>

algorithm_C = np.array([45, 43, 23, 49, 49, 8, 21, 20, 42, 40, 28, 46, 44, 37, 44, 38, 42, 34, 42, 40])<br/>

Check for the assumptions of One-way ANOVA & decide the appropriate test to determine if there is a statistically significant difference in the execution times of the three algorithms at a 95% confidence level.

```python
H0 = 'Execution times are same'
HA = 'Execution times are different'
alpha = 0.05
algorithm_A = np.array([23, 25, 22, 27, 28, 24, 26, 29, 21, 30, 25, 43, 26, 28, 24, 22, 27, 46, 25, 29]) 
algorithm_B = np.array([31, 28, 29, 32, 30, 33, 27, 28, 32, 30, 31, 29, 30, 48, 33, 31, 29, 30, 32, 31]) 
algorithm_C = np.array([45, 43, 23, 49, 49, 8, 21, 20, 42, 40, 28, 46, 44, 37, 44, 38, 42, 34, 42, 40])

s_test, p_value = shapiro(algorithm_A)
if p_value < alpha:
    print('algorithm_A is not normal distribution')
else:
    print('algorithm_A is normal distribution')

s_test, p_value = shapiro(algorithm_B)
if p_value < alpha:
    print('algorithm_B is not normal distribution')
else:
    print('algorithm_B is normal distribution')    

s_test, p_value = shapiro(algorithm_C)
if p_value < alpha:
    print('algorithm_C is not normal distribution')
else:
    print('algorithm_C is normal distribution')   

f_test, p_value = kruskal(algorithm_A, algorithm_B,algorithm_C)
print(f'F_One way stat: {f_stat}')
print(f'p_value is: {p_value}, 9 decimals: {p_value:.9f}')

if p_value < alpha:
    print(f'Reject H0 means: {HA}')
else:
    print(f'Fail to reject H0 means: {H0}')
```

### Q8. Different Variance? <br/>
The pharmaceutical company collected data on the treatment effects (values in corresponding units) of two drug formulations (Group A and Group B) in a study<br/>

group_a_effects = [1.5, 1.8, 1.2, 1.6, 1.7]<br/>
group_b_effects = [1.4, 1.9, 1.3, 1.5, 1.8]<br/>

The researchers want to assess whether the variances of the treatment effects are significantly different among the two groups.<br/>
Perform the appropriate test with a 90% confidence level, calculate the p-value and interpret the conclusions.

```python
H0 = 'Variances are similar'
HA = 'Variances are different'
group_a_effects = [1.5, 1.8, 1.2, 1.6, 1.7]
group_b_effects = [1.4, 1.9, 1.3, 1.5, 1.8]
alpha = 0.1

levene_stat, p_value = levene(group_a_effects,group_b_effects)
print(f'levene stat: {levene_stat} p_value: {p_value:.8f}')

if p_value < alpha:
    print(f'Reject NULL Hypothesis means: {HA}')
else:
    print(f'Fail to reject NULL Hypothesis means: {H0}')
```
