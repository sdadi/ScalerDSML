```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

### Q1. 
A company wants to estimate the average purchase amount made by customers on their e-commerce platform. They collect a random sample of 100 transactions and found that the average purchase amount is  75withastandarddeviationof 10. Construct a 90% confidence interval for the true average purchase amount

```python
sample = 100
mean = 75
std_dev = 10
std_err = std_dev / np.sqrt(sample)
confidence_percent = 90
confidence_level = confidence_percent/100

z1 = norm.ppf(0+(1-confidence_level)/2)
z2 = norm.ppf(confidence_level+ (1-confidence_level)/2)
print(f"The z scores are {z1} and {z2}")

x1 = mean + z1*std_err
x2 = mean + z2*std_err
print(f"The {confidence_percent}% confidence interval is between {x1} and {x2}")

# we can use norm.interval(loc=mean, scale=std_dev/np.sqrt(sample), confidence=confidence_level) directly also 
confidence_interval = norm.interval(confidence=confidence_level, loc=mean, scale=std_err)
confidence_interval
```

### Q2 
The tip percentage for a delivery person has a mean value of 28% and a standard deviation of 7%.

What is the probability that the sample mean tip percentage for a random sample of 50 bills is between 26% and 31%?

```python
std_dev = 7
mean = 28
sample=50
x1=26
x2=31

std_err = std_dev/np.sqrt(sample)

z_lower = (x1 - mean)/std_err
z_upper = (x2 - mean)/std_err
print(f"The z scores are {z_lower} and {z_upper}")

norm.cdf(z_upper) - norm.cdf(z_lower)

```

###Q3
You are working for the Department of Endangered Species. Along with your team, you have bred a certain species of endangered fish in your pond.

On taking a sample of 100 fishes, the calculated average length of this species came out to be 24cm. It is known that the population standard deviation in the length of these fishes is 8cm.

Estimate the mean length confidence interval of the fishes in the pond with a confidence of 95 percent.

```python
sample = 100
mean = 24
std_dev = 8
confidence_percent = 95
confidence_level = confidence_percent/100

std_err = std_dev / np.sqrt(sample)

z1 = norm.ppf(0+(1-confidence_level)/2)
z2 = norm.ppf(confidence_level+ (1-confidence_level)/2)
print(f"The z scores are {z1} and {z2}")

x1 = mean + z1*std_err
x2 = mean + z2*std_err
print(f"The {confidence_percent}% confidence interval is between {x1} and {x2}")

norm.interval(loc=mean, scale=std_err, confidence=confidence_level)
```

### Q4
An unknown distribution has a mean of 90 and a standard deviation of 15.

Samples of size n = 35 are drawn randomly from the population.

Find the probability that the sample mean is between 85 and 92.

```python
mean = 90
std_dev = 15
sample = 35
x1 = 85
x2 = 92

std_err = std_dev/np.sqrt(sample)

z1 = (85-mean)/std_err
z2 = (92-mean)/std_err
print(f"The z scores are {z1} and {z2}")

norm.cdf(z2) - norm.cdf(z1)
```
