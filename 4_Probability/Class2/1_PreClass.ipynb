Q3: find out how many matches India have won when Sachin has scored a century

```python
my_pivot = pd.crosstab(index=df_sachin['century'], columns=df_sachin['Won'],margins=True)
my_pivot
```

```python
# Get matches India won when Sachin didn't score a century using my_pivot
wins_without_century_from_pivot = my_pivot.loc[False, True]
print("Matches India won when Sachin didn't score a century (from my_pivot):", wins_without_century_from_pivot)
```

```python
# Matches India won when Sachin scored a century using my_pivot
wins_with_century_from_pivot = my_pivot.loc[True, True]
print("Matches India won when Sachin scored a century (from my_pivot):", int(wins_with_century_from_pivot))
```

```python
# Count matches India won when Sachin scored a century
count = df_sachin.loc[(df_sachin['century']) & (df_sachin['Won'])].shape[0]
print("Matches India won when Sachin scored a century:", count)
```

Q4. How many matches India have won when sachin didn't score a century.

```python
# Matches India won when Sachin didn't score a century
wins_without_century = df_sachin.loc[(~df_sachin['century']) & (df_sachin['Won'])].shape[0]
print("Matches India won when Sachin didn't score a century:", wins_without_century)
```
