# Q3. Given that you have already spent 3 minutes, what is the probability of needing more than 9 minutes

```python
# P[T > 9 | T > 3]= P[T>9]/P[T>3] =P(T>6)

(1-expon.cdf(9,scale=5))/(1-expon.cdf(3,scale=5))
```

The memoryless property essentially means that the time you've already spent on an event doesn't affect the future waiting time.
In the context of the exponential distribution, it means that the probability of needing more time in the future is the same, regardless of how much time has already passed.

A food delivery app records the time taken for deliveries in minutes:

50% of the orders are delivered in 25 minutes.

30% take 35 minutes.

20% take 45 minutes.

Question 1: What is the expected delivery time?

Question 2: If a customer places 3 independent orders, what is the expected total delivery time?

Options:

A. Expected time = 30 minutes, Total time = 90 minutes

B. Expected time = 32 minutes, Total time = 96 minutes

C. Expected time = 35 minutes, Total time = 105 minutes

D. Expected time = 40 minutes, Total time = 120 minutes

```python
# Given data
times = np.array([25, 35, 45])  # Delivery times in minutes
probabilities = np.array([0.5, 0.3, 0.2])  # Probabilities

# 1️⃣ Expected delivery time (E(X) = ∑x_i * P(x_i))
expected_delivery_time = np.sum(times * probabilities)

# 2️⃣ Expected delivery time for 3 orders (Linearity: E(3X) = 3 * E(X))
expected_total_time_3_orders = 3 * expected_delivery_time

print(f"Expected delivery time for one order: {expected_delivery_time:.2f} minutes")
print(f"Expected total delivery time for 3 orders: {expected_total_time_3_orders:.2f} minutes")
```
