# You are working as a data engineer who has to resolve any bugs
# failures of ML models in production.
# The time taken to debug is exponentially distributed with mean of 5 minutes
# Find the probability of debugging in 4 to 5 minutes?

```python
# 1 bug -> 5 minutes
# x bugs -> 1 minutes
# x(rate) -> 1/5

rate = 1/5
scale = 1/rate
# scale = 5

# P([4<X<5]) = CDF(5) - CDF(4)
expon.cdf(5,scale=scale) - expon.cdf(4,scale=scale)
```

# Find the probability of needing more than 6 minutes to debug

```python
1-expon.cdf(6,scale=scale)
```

# Given that you have already spent 3 minutes, what is the probability of needing more than 9 minutes

```python
# P[T>9|T>3] = P[(T>9) and (T>3)]/P(T>3) = P(T>9)/P(T>3)
(1-expon.cdf(9,scale=scale))/(1-expon.cdf(3,scale=scale))

#  same as P[T>6]
1-expon.cdf(6,scale=scale)
```

# Suppose you are playing a game where success rate of winning a prize is 0.7.
# (a) What is the probability of winning a prize on 4th attempt?
# (b) What is the probability that you don't win in 2 attempts?
# (c) What is the expected number of trials to get the first success?

```python
a = geom.pmf(k=4, p=0.7)
print(f'Prob of winning on 4th attempt is {a}')

b = 1 - geom.cdf(k=2, p=0.7)
print(f'Prob of not winning on 2 attempts is {b}')

c = 1/0.7
print(f'Expected number of attempts to win is {c}')
```
