A call centre receives 20 calls per hour, on an average.

a) What is the probability of receiving exactly 25 calls in a given hour?

b) What is the probability of receiving more than 25 calls in a given hour?

```python
mu=20
k=25
# P(X=25)
q1 = poisson.pmf(k=25,mu=20)
print(f'prob of exactly 25 in hour {q1}')

q2 = 1-poisson.cdf(k=25,mu=20)
print(f'prob of receiving more than 25 calls in hour {q2}')
```

You are hired as a data scientist for the company SellMart.

Your first task is to analyze the durations of customer support call.

On an average, a customer care executive takes about 5 minutes to solve the issue of any customer. It is known that the time taken follows an exponential distribution.

Find the probability that an executive spends 4 to 5 minutes with any randomly selected customer.

```python
mu=5

q1 = expon.cdf(x=5,scale=mu) - expon.cdf(x=4,scale=mu)
q1
```

It is known that 20% of Netflix users are premium subscribers.

Also, 80% of premium subscribers watch at least one thriller movie per week, and 50% of basic subscribers watch at least one thriller movie per week.

What is the probability that a Netflix subscriber who watches at least one thriller movie per week is a premium subscriber?

```python
p_s = .2 # P(Prem)
n_s =.8  # P(Basic)

p_thriller =0.8  #P(T | Prem)
n_thriller = 0.5 #P(T | Basic)

# p_thriller = 0.8*p_s 
# n_thriller = 0.5*n_s 

thriller = (p_thriller*p_s+n_thriller*n_s)

q1 = p_thriller*p_s/thriller # P(Prem | T)
print(f'probability of prem subs watching is {q1}')

#  we can use Bayes theorem when P(A | B) is given find P(B | A) 
#  P(B | A) = P(A|B)*P(A)/P(B)
q1 = (p_thriller*p_s)/thriller
q1

```
